```python
import numpy as np
import pandas as pd
import pickle
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Load the dataset using Pandas
df = pd.read_csv("/content/diamonds.csv")
```

```python
# Define the input features (X) and the target variable (price)
X = df.drop(columns=["price"])
y = df["price"]
```

```python
print("Input features (X):")
print(X.head())
print("\nTarget variable (y):")
print(y.head())
```

```python
# Split the dataset into training and testing sets using a 75:25 ratio
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
print("Training set shape:")
print(X_train.shape, y_train.shape)
print("\nTesting set shape:")
print(X_test.shape, y_test.shape)
```

```python
# Apply data preprocessing (Encode categorical features & Scale numerical features)
numerical_features = ["carat", "depth", "table", "x", "y", "z"]
categorical_features = ["cut", "color", "clarity"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ]
)
```

```python
print("Preprocessor:")
print(preprocessor)
```

```python
# Train a KNN Regressor model using Scikit-learn Pipeline
model_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", KNeighborsRegressor(n_neighbors=5)),
    ]
)

print("Training the model...")
model_pipeline.fit(X_train, y_train)
```

```python
# Evaluate the model
y_pred = model_pipeline.predict(X_test)
print(f"MAE: {mean_absolute_error(y_test, y_pred):.2f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.2f}")
print(f"R2 Score: {r2_score(y_test, y_pred):.4f}")
```

```python
# Save the trained pipeline using Pickle
with open("model.pkl", "wb") as file:
  pickle.dump(model_pipeline, file)
print("Pipeline saved successfully as model.pkl!")
```
